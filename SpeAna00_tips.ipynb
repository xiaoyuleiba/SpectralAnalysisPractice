# <span style="color:DeepPink;"> How to Convert Notebook Files</span>

Jupyter Notebook（`.ipynb`）は、保存や共有のためにさまざまな形式へ変換できます。  
このセクションでは、次の変換方法を紹介します：

- `.py` ファイル：Python スクリプトとして保存（コードのみ）  
- `.html` ファイル：Web ページ形式に変換  
- `.pdf` ファイル：印刷や配布用に PDF として出力  

具体的なコマンド例は以下のコードセルに示しています。

### <span style="color:SeaGreen;">変換方法</span>

### 1. Python ファイル（`.py`）へ変換する

ノートブックからコードセルだけを取り出して `.py` ファイルとして保存できます。  
これにより、Jupyter 以外の VS Code やその他エディタでもコードを実行できます。

- Markdown セルはコメントとして残ります  
- 実行結果や出力は含まれません  
---

### 2. HTML ファイル（`.html`）へ変換する

ノートブックを Web ページのような形式に変換できます。  
見出し・コード・出力がすべて含まれ、元のレイアウトが保たれます。

- オンラインドキュメントやレポートとして共有する際に便利です

---

### 3. PDF ファイル（`.pdf`）へ変換する

ノートブックを PDF として書き出すことができます。  
ただし LaTeX 環境が必要な場合があります。  
もし PDF 変換が失敗した場合は、HTML に変換してから PDF に印刷する方法がおすすめです。

---

変換は通常、ターミナルまたは Jupyter のコードセルから実行します。  
以下はコマンド例です：

```python
# .ipynb → .py（Pythonスクリプトのみ）
!jupyter nbconvert --to script your_notebook.ipynb

# .ipynb → .html（レイアウトを完全保持）
!jupyter nbconvert --to html your_notebook.ipynb

# .ipynb → .pdf（LaTeX環境が必要）
!jupyter nbconvert --to pdf your_notebook.ipynb

# コード入力を除いたHTML（配布資料向け）
!jupyter nbconvert --to html --no-input your_notebook.ipynb
```

### <span style="color:SeaGreen;">.py ファイルの実行方法</span>

ノートブックを `.py` ファイルに変換したら、次は Python プログラムとして実行します。
